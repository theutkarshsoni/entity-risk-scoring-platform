### `dim_entity`

entity_id | UUID (PK) | System-generated

entity_type | string | `human` / `service_account` / `agent`
- 68 human / 21 service_account / 11 agent

role | string | Human: Finance, Engineering, IT/Admin, HR, Sales. Service_account/agent: function (e.g. data-pipeline, backup-automation, monitoring, ci-cd, security-scanning, ai-assistant, integration)
- weighted-but-distributed within each type

tier | string | Human: Junior/Senior/Manager. Non-human: Supervised/Semi-autonomous/Fully-autonomous — acts as a *ceiling* on access grants, not a direct grant driver
- human — Manager < Junior/Senior (pyramid); non-human — roughly even, not perfectly uniform

activity_pattern | string, nullable | Service_account only: scheduled / triggered / always_on. Null for human and agent
- random, skewed toward scheduled
- More scheduled activity_pattern to make anomaly detection actually testable — good, deliberate reasoning, not arbitrary

home_country | string, nullable | Human only. Null for service_account and agent
- within Europe, human only

created_at | timestamp 

```python
# List of 44 European countries
europe_countries = [
    "Albania", "Andorra", "Austria", "Belarus", "Belgium", 
    "Bosnia and Herzegovina", "Bulgaria", "Croatia", "Cyprus", "Czechia", 
    "Denmark", "Estonia", "Finland", "France", "Germany", 
    "Greece", "Hungary", "Iceland", "Ireland", "Italy", 
    "Latvia", "Liechtenstein", "Lithuania", "Luxembourg", "Malta", 
    "Moldova", "Monaco", "Montenegro", "Netherlands", "North Macedonia", 
    "Norway", "Poland", "Portugal", "Romania", "Russia", 
    "San Marino", "Serbia", "Slovakia", "Slovenia", "Spain", 
    "Sweden", "Switzerland", "Ukraine", "United Kingdom"
]

# Specs of activity pattern
activity_patterns  = ["scheduled", "triggered",  "always_on"]
activity_weights = [0.5, 0.3, 0.2]

# Specs of role
human_roles = ["Finance", "Engineering", "IT", "HR", "Sales"]
human_role_weights = [0.20, 0.35, 0.20, 0.10, 0.15]
non_human_roles = ["data-pipeline", "backup-automation", "monitoring", "ci-cd", "security-scanning", "ai-assistant", "integration"]
non_human_role_weights = [0.20, 0.10, 0.10, 0.15, 0.15, 0.15, 0.15]

# Specs of tier
human_tiers = ["Junior", "Senior", "Manager"]
human_tier_weights = [0.40, 0.35, 0.25]
non_human_tiers = ["Supervised", "Semi-autonomous", "Fully-autonomous"]
non_human_tier_weights = [0.30, 0.30, 0.40]

import random
```

```python
new_entity_list = []
```

```python
# generate human
e_type = "human"
for i in range(6): # it should be 68 when script is finalised
    human_entity = {}
    human_entity["entity_type"] = e_type
    role = random.choices(human_roles, weights=human_role_weights, k=1)
    human_entity["role"] = role[0]
    tier = random.choices(human_tiers, weights=human_tier_weights, k=1)
    human_entity["tier"] = tier[0]
    country = random.choice(europe_countries)
    human_entity["home_country"] = country
    new_entity_list.append(human_entity)
```

```python
# generate service account
e_type = "service_account"
for j in range (2): # it should be 21 when script is finalised
    service_entity = {}
    service_entity["entity_type"] = e_type
    role = random.choices(non_human_roles, weights=non_human_role_weights, k=1)
    service_entity["role"] = role[0]
    tier = random.choices(non_human_tiers, weights=non_human_tier_weights, k=1)
    service_entity["tier"] = tier[0]
    pattern = random.choices(activity_patterns, weights=activity_weights, k=1) # The "Loaded Dice" Approach
    service_entity["activity_pattern"] = pattern[0]
    new_entity_list.append(service_entity)
```

```python
# generate agent
e_type = "agent"
for k in range(1): # it should be 11 when script is finalised
    agent_entity = {}
    agent_entity["entity_type"] = e_type
    role = random.choices(non_human_roles, weights=non_human_role_weights, k=1)
    agent_entity["role"] = role[0]
    tier = random.choices(non_human_tiers, weights=non_human_tier_weights, k=1)
    agent_entity["tier"] = tier[0]
    new_entity_list.append(agent_entity)
```
